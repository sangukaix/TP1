# 04. 결과 해석(Interpretation)

통계적 연관성 결과와 머신러닝 결과를 발표에서 설명할 수 있도록 정리합니다.


```python
import pandas as pd
from pathlib import Path

ART = Path('model_artifacts/asd')
assoc = pd.read_csv(ART/'all_feature_association.csv')
importance = pd.read_csv(ART/'selected_feature_importance.csv')
models = pd.read_csv(ART/'model_validation_comparison.csv')
final = pd.read_csv(ART/'final_candidate_results.csv')

related = assoc[assoc['significant_0_05'] == True]
not_related = assoc[assoc['significant_0_05'] == False]

print('관련성 확인 요인:', related['feature'].tolist())
print('뚜렷한 관련성 확인 안 된 요인:', not_related['feature'].tolist())
display(models[['model','accuracy','precision','recall','f1','roc_auc']])
display(final)
```

## 해석 포인트
- A1~A10은 모두 `Class/ASD`와 관련성이 확인됩니다. 다만 `Class/ASD` 자체가 A1~A10 합계 규칙으로 생성되므로 이는 어느 정도 예상되는 결과입니다.
- 개인·배경 요인은 p-value와 연관성 크기를 함께 보고, 희소 범주가 많은 변수는 해석에 주의합니다.
- 머신러닝에서는 3번 연관성 분석에서 p < 0.05인 요인을 입력 후보로 사용합니다.
- 높은 모델 성능을 실제 임상 진단 정확도라고 해석하지 않습니다.


## 프로젝트 활용 문장

**이 프로젝트는 ASD 선별 관련 행동을 이용해 아동학대를 직접 예측하는 것이 아닙니다. ASD 선별 관련 행동이 반복되는 아동을 추가 관찰 대상으로 보고, 취약성을 고려하여 별도의 학대 의심 징후를 조기에 놓치지 않도록 돕는 관찰·연계 보조 자료를 만드는 것이 목적입니다.**
